Welcome! This notebook demonstrates how to load macroeconomic data, compute 2D business cycle features ($X$ Health, $Y$ Momentum), and generate quantitative projections using the Python API.

```python
import sys
from pathlib import Path

# Locate the repository whether Jupyter starts in the project root, notebooks/,
# or the parent folder created by a GitHub ZIP download.
for candidate in (Path.cwd(), *Path.cwd().parents):
    roots = [candidate, *(path.parent for path in candidate.glob('*/macro_intel.py'))]
    REPO_ROOT = next((root for root in roots if (root / 'macro_intel.py').is_file()), None)
    if REPO_ROOT is not None:
        break
else:
    raise RuntimeError('Open this notebook from the macro_intelligence_platform repository or its parent folder.')

if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

from macro_intel import load_macro_data, compute_features, forecast_cycle
import pandas as pd
```

## 1. Load Macro Data (`DataBundle`)

Load indicator, macro driver, and market context series. In offline mode, uses bundled local fallbacks.

```python
# Choose either 'INDIA' or 'US'. Offline mode uses bundled/cache data when available.
MARKET = 'INDIA'
bundle = load_macro_data(market=MARKET, offline=True)
print(f"Data As-Of: {bundle.as_of}")
print("\nData Provenance & Health Metadata:")
for k, v in list(bundle.data_health.items())[:5]:
    print(f"  {k}: {v['source']} ({v['cache_status']})")
```

## 2. Compute Features ($X$ Health, $Y$ Momentum, Quadrants)

```python
bundle = compute_features(bundle)
bundle.df[['X', 'Y', 'Quadrant']].tail(10)
```

## 3. Project Business Cycle Forward (`ForecastResult`)

```python
result = forecast_cycle(bundle)
print(f"Current Regime: {result.current_regime}")
print(f"Headline 6M Forecast: {result.forecasts['6m'].quadrant} (Conviction: {result.forecasts['6m'].conviction}%)")
print(f"Model Version: {result.model_version}")
```
